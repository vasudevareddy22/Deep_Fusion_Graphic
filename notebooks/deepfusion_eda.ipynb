## 2. Graph Topology Construction
Creating a directed graph where entities are nodes and traffic sessions are edges.

```python
G = nx.DiGraph()
for _, row in df.iterrows():
    G.add_edge(row['src_ip'], row['dst_ip'], weight=row['count'])

print(f"Network Entities (Nodes): {G.number_of_nodes()}")
print(f"Communication Channels (Edges): {G.number_of_edges()}")
```
